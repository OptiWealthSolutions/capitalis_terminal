```python
import pandas_datareader.data as web
import pandas as pd
from datetime import datetime, timedelta

start = datetime.now() - timedelta(days=5*365)
end = datetime.now()

tickers = {
    'CPIAUCSL': 'CPI',
    'INDPRO': 'PMI', 
    'UNRATE': 'Unemployment',
    'FEDFUNDS': 'Interest_Rate'
}

df_source = web.DataReader(list(tickers.keys()), 'fred', start, end)
df_source = df_source.rename(columns=tickers).resample('ME').last().ffill()

df_final = pd.DataFrame(index=df_source.index)

for col in df_source.columns:
    df_final[f'{col}_Actual'] = df_source[col]
    df_final[f'{col}_Consensus'] = df_source[col].shift(1)

df_final = df_final.dropna()
```
